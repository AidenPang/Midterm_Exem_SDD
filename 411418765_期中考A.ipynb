````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
DB_NAME = "campus_events.db"

def init_database():
    """初始化資料庫：重建資料表並寫入活動初始資料"""
    conn = sqlite3.connect(DB_NAME)
    try:
        cursor = conn.cursor()

        # 1. 先刪除舊資料表（若存在）
        cursor.execute("DROP TABLE IF EXISTS registrations")
        cursor.execute("DROP TABLE IF EXISTS activities")

        # 2. 建立 activities（活動表）
        cursor.execute("""
            CREATE TABLE activities (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT UNIQUE NOT NULL,
                date TEXT NOT NULL,
                capacity INTEGER NOT NULL
            )
        """)

        # 3. 建立 registrations（報名紀錄表）
        cursor.execute("""
            CREATE TABLE registrations (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                activity_id INTEGER NOT NULL,
                student_name TEXT NOT NULL,
                party_size INTEGER NOT NULL,
                status TEXT NOT NULL,
                FOREIGN KEY (activity_id) REFERENCES activities (id)
            )
        """)

        # 4. 插入題目規定的 3 場初始活動資料（全部使用 ? 參數防護）
        initial_data = [
            ("迎新晚會", "2026-11-20", 6),
            ("程式工作坊", "2026-11-25", 4),
            ("籃球友誼賽", "2026-11-28", 3)
        ]
        cursor.executemany(
            "INSERT INTO activities (name, date, capacity) VALUES (?, ?, ?)",
            initial_data
        )

        conn.commit()
        print("✅ 資料庫初始化完成，已寫入 3 場活動初始資料。")
    finally:
        # 確保連線一定會關閉
        conn.close()

# 執行初始化
init_database()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def show_tables():
    """查詢並顯示活動表與報名紀錄表的所有資料"""
    conn = sqlite3.connect(DB_NAME)
    try:
        print("📌 【活動資料表 (activities)】")
        df_activities = pd.read_sql_query("SELECT * FROM activities", conn)
        display(df_activities)

        print("\n📌 【報名紀錄表 (registrations)】")
        df_registrations = pd.read_sql_query("SELECT * FROM registrations", conn)
        if df_registrations.empty:
            print("（目前尚無任何報名紀錄）")
        else:
            display(df_registrations)
    finally:
        conn.close()

# 顯示目前資料表狀態
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# 定義允許清單（白名單）
ALLOWED_ACTIONS = ["報名", "取消", "查詢", "統計"]
ALLOWED_ACTIVITIES = ["迎新晚會", "程式工作坊", "籃球友誼賽"]

# 暫存字典：同一句話不重複呼叫 AI
ai_cache = {}

def call_groq_with_retry(prompt_text):
    """呼叫 Groq API，遇 429 錯誤最多重試 3 次，每次呼叫後暫停 2 秒"""
    max_retries = 3
    for attempt in range(max_retries):
        try:
            chat_completion = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                reasoning_effort=REASONING_EFFORT,
                messages=[
                    {"role": "user", "content": prompt_text}
                ]
            )
            time.sleep(2)  # 每次呼叫後休息 2 秒，避免額度超標
            return chat_completion.choices[0].message.content.strip()
        except Exception as e:
            error_message = str(e)
            if "429" in error_message and attempt < max_retries - 1:
                print(f"⚠️ 遇到使用量上限 (429)，等待 30 秒後進行第 {attempt + 2} 次重試...")
                time.sleep(30)
            else:
                print(f"❌ 呼叫 AI 時發生錯誤：{error_message}")
                return "未知,無,1"
    return "未知,無,1"

def parse_user_request(user_text):
    """
    輸入使用者語句，回傳解讀結果 (動作, 活動, 人數, 原始回覆)。
    依據規定嚴格驗證 AI 回覆，不合規則者給予預設值。
    """
    cleaned_text = user_text.strip()

    # 檢查暫存
    if cleaned_text in ai_cache:
        return ai_cache[cleaned_text]

    prompt = f"""你是一個精確的系學會活動報名助理。請閱讀使用者的這句話，並依指定格式輸出 3 個欄位：動作,活動,人數

嚴格遵守以下規則：
1. 動作判定標準（只能是以下五者之一）：
   - 報名：想參加、報名、登記活動
   - 取消：想退選、退出、不去了、取消活動
   - 查詢：詢問特定「單一活動」的剩餘名額或狀況
   - 統計：詢問「各活動、所有活動、全部活動」的報名狀況、統計清單或總覽
   - 未知：無法理解或與活動無關的事務
2. 活動只能是：迎新晚會、程式工作坊、籃球友誼賽、無（若動作為「統計」或未指明特定活動，請填「無」）
3. 人數只能是阿拉伯數字（例如 1、2、3），預設為 1。若有攜伴請加總（例如「我和兩位朋友」輸出 3）。取消、查詢或統計若無提及人數請填 1。
4. 只能輸出格式：動作,活動,人數，不要有任何 Markdown 或其他說明文字。

使用者輸入："{cleaned_text}"
"""
    raw_reply = call_groq_with_retry(prompt)

    # --- 驗證與萃取（不直接相信 AI） ---

    # 1. 動作比對（白名單）
    action = "未知"
    for allowed_act in ALLOWED_ACTIONS:
        if allowed_act in raw_reply:
            action = allowed_act
            break

    # 2. 活動比對（白名單）
    activity = "無"
    for allowed_event in ALLOWED_ACTIVITIES:
        if allowed_event in raw_reply:
            activity = allowed_event
            break

    # 3. 數量擷取（使用正規表示式找出數字，找不到就預設為 1）
    found_numbers = re.findall(r"\d+", raw_reply)
    if found_numbers:
        party_size = int(found_numbers[0])
    else:
        party_size = 1

    result = (action, activity, party_size, raw_reply)
    ai_cache[cleaned_text] = result
    return result

print("✅ AI 解讀模組載入完成。")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# 測試資料清單
test_cases = [
    ("我想報名迎新晚會", "報名", "迎新晚會", 1),
    ("我要參加程式工作坊，我和兩個朋友", "報名", "程式工作坊", 3),
    ("幫我取消籃球友誼賽", "取消", "籃球友誼賽", 1),
    ("請問迎新晚會還有名額嗎", "查詢", "迎新晚會", 1),
    ("各活動報名狀況如何", "統計", "*", "*"),
    ("我想帶4個朋友去迎新晚會", "報名", "迎新晚會", 5),
    ("退選程式工作坊", "取消", "程式工作坊", 1),
    ("明天會下雨嗎？", "未知", "*", "*"),
    ("籃球友誼賽幾點開始？剩幾個人？", "查詢", "籃球友誼賽", 1)
]

passed_count = 0
total_cases = len(test_cases)

print(f"🚀 開始執行 AI 解讀測試（共 {total_cases} 句）...\n" + "-" * 60)

for idx, (sentence, exp_action, exp_activity, exp_count) in enumerate(test_cases, 1):
    action, activity, count, raw_reply = parse_user_request(sentence)

    # 驗證是否符合預期（"*" 代表不拘）
    action_match = (exp_action == "*") or (action == exp_action)
    activity_match = (exp_activity == "*") or (activity == exp_activity)
    count_match = (exp_count == "*") or (count == exp_count)

    is_success = action_match and activity_match and count_match
    if is_success:
        passed_count += 1
        result_symbol = "✅ 通過"
    else:
        result_symbol = "❌ 失敗"

    print(f"測試 {idx} [{result_symbol}]：{sentence}")
    print(f"   AI 原始回覆 ：{raw_reply}")
    print(f"   系統解讀結果：動作={action}, 對象={activity}, 數量={count}")
    print(f"   預期解讀結果：動作={exp_action}, 對象={exp_activity}, 數量={exp_count}\n")

print("-" * 60)
print(f"🎯 測試統計：共測試 {total_cases} 句，成功通過 {passed_count} 句！")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

def get_activities_statistics():
    """A3 進階需求：使用 JOIN 與 GROUP BY 統計各活動，未有人報名也必須列出"""
    conn = sqlite3.connect(DB_NAME)
    try:
        cursor = conn.cursor()
        # 使用 LEFT JOIN 與 GROUP BY，計算已報名人數與剩餘名額
        sql = """
            SELECT
                a.name AS "活動名稱",
                a.date AS "日期",
                a.capacity AS "總名額",
                COALESCE(SUM(CASE WHEN r.status = 'REGISTERED' THEN r.party_size ELSE 0 END), 0) AS "已報名人數",
                (a.capacity - COALESCE(SUM(CASE WHEN r.status = 'REGISTERED' THEN r.party_size ELSE 0 END), 0)) AS "剩餘名額"
            FROM activities a
            LEFT JOIN registrations r ON a.id = r.activity_id
            GROUP BY a.id, a.name, a.date, a.capacity
            ORDER BY a.date ASC
        """
        cursor.execute(sql)
        rows = cursor.fetchall()

        df = pd.DataFrame(rows, columns=["活動名稱", "日期", "總名額", "已報名人數", "剩餘名額"])
        display(df)
        table_text = df.to_string(index=False)
        return f"📊 【各活動報名狀況統計】\n\n{table_text}"
    finally:
        conn.close()

def query_event(activity_name):
    """查詢單一活動剩餘名額（不需要姓名）"""
    conn = sqlite3.connect(DB_NAME)
    try:
        cursor = conn.cursor()
        cursor.execute("SELECT id, capacity FROM activities WHERE name = ?", (activity_name,))
        act_row = cursor.fetchone()
        if act_row is None:
            return f"❌ 找不到活動「{activity_name}」。"

        act_id, capacity = act_row
        cursor.execute("""
            SELECT COALESCE(SUM(party_size), 0)
            FROM registrations
            WHERE activity_id = ? AND status = 'REGISTERED'
        """, (act_id,))
        booked_row = cursor.fetchone()
        booked = booked_row[0] if booked_row is not None else 0
        remaining = capacity - booked
        return f"ℹ️ 【{activity_name}】總名額 {capacity} 人，已報名 {booked} 人，目前剩餘名額：{remaining} 人。"
    finally:
        conn.close()

def register_event(student_name, activity_name, party_size):
    """
    報名活動處理函式：
    - 姓名檢查
    - A1: 人數必須介於 1~3 人之間
    - 重複報名同一活動檢查
    - A2: 每位學生最多同時報名 2 個活動
    - 剩餘名額檢查
    """
    if not student_name:
        return "❌ 報名失敗：必須提供學生姓名！"

    # A1 規範檢查：單次報名人數必須介於 1 到 3 之間
    if not (1 <= party_size <= 3):
        return f"❌ 報名失敗：單次報名人數限制為 1～3 人（您輸入了 {party_size} 人）。"

    conn = sqlite3.connect(DB_NAME)
    try:
        cursor = conn.cursor()

        # 1. 檢查活動是否存在
        cursor.execute("SELECT id, capacity FROM activities WHERE name = ?", (activity_name,))
        act_row = cursor.fetchone()
        if act_row is None:
            return f"❌ 找不到活動「{activity_name}」。"
        act_id, capacity = act_row

        # 2. 檢查是否重複報名同一個活動（有效狀態下不可重複）
        cursor.execute("""
            SELECT id FROM registrations
            WHERE student_name = ? AND activity_id = ? AND status = 'REGISTERED'
        """, (student_name, act_id))
        if cursor.fetchone() is not None:
            return f"❌ 報名失敗：【{student_name}】已經報名過「{activity_name}」，不可重複報名！"

        # 3. A2 規範檢查：每人最多同時報名 2 個活動（已取消的不計）
        cursor.execute("""
            SELECT COUNT(*) FROM registrations
            WHERE student_name = ? AND status = 'REGISTERED'
        """, (student_name,))
        count_row = cursor.fetchone()
        active_count = count_row[0] if count_row is not None else 0
        if active_count >= 2:
            return f"❌ 報名失敗：每人最多同時報名 2 個活動，【{student_name}】已累計報名 {active_count} 個活動。"

        # 4. 檢查名額是否足夠
        cursor.execute("""
            SELECT COALESCE(SUM(party_size), 0) FROM registrations
            WHERE activity_id = ? AND status = 'REGISTERED'
        """, (act_id,))
        booked_row = cursor.fetchone()
        booked = booked_row[0] if booked_row is not None else 0
        remaining = capacity - booked

        if party_size > remaining:
            return f"❌ 報名失敗：「{activity_name}」剩餘名額不足（剩餘 {remaining} 人，欲報名 {party_size} 人）。"

        # 5. 寫入報名資料
        cursor.execute("""
            INSERT INTO registrations (activity_id, student_name, party_size, status)
            VALUES (?, ?, ?, 'REGISTERED')
        """, (act_id, student_name, party_size))
        conn.commit()
        return f"🎉 報名成功！【{student_name}】已成功報名「{activity_name}」，共 {party_size} 位名額。"
    finally:
        conn.close()

def cancel_event(student_name, activity_name):
    """取消報名處理函式：查出報名紀錄並將狀態更新為 CANCELLED，釋出名額"""
    if not student_name:
        return "❌ 取消失敗：必須提供學生姓名！"

    conn = sqlite3.connect(DB_NAME)
    try:
        cursor = conn.cursor()

        # 1. 取得活動 ID
        cursor.execute("SELECT id FROM activities WHERE name = ?", (activity_name,))
        act_row = cursor.fetchone()
        if act_row is None:
            return f"❌ 找不到活動「{activity_name}」。"
        act_id = act_row[0]

        # 2. 尋找該學生是否有該活動的有效報名紀錄
        cursor.execute("""
            SELECT id, party_size FROM registrations
            WHERE student_name = ? AND activity_id = ? AND status = 'REGISTERED'
        """, (student_name, act_id))
        reg_row = cursor.fetchone()
        if reg_row is None:
            return f"⚠️ 取消失敗：查無【{student_name}】報名「{activity_name}」的有效紀錄。"

        reg_id, p_size = reg_row

        # 3. 更新狀態為 CANCELLED
        cursor.execute("UPDATE registrations SET status = 'CANCELLED' WHERE id = ?", (reg_id,))
        conn.commit()
        return f"✅ 取消成功：已為【{student_name}】取消「{activity_name}」報名，釋出 {p_size} 個名額。"
    finally:
        conn.close()

print("✅ 系統業務功能函式載入完成。")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

def process_system_request(student_name, user_text):
    """主流程派發函式"""
    cleaned_name = student_name.strip() if student_name else ""
    cleaned_text = user_text.strip() if user_text else ""

    if not cleaned_text:
        return "⚠️ 請輸入您的需求文字！"

    action, activity, count, _ = parse_user_request(cleaned_text)

    # 1. 統計查詢
    if action == "統計":
        return get_activities_statistics()

    # 2. 活動無法辨識時，提示目前有哪些活動
    if activity not in ALLOWED_ACTIVITIES:
        conn = sqlite3.connect(DB_NAME)
        try:
            cursor = conn.cursor()
            cursor.execute("SELECT name, date, capacity FROM activities")
            rows = cursor.fetchall()
            list_text = "\n".join([f"• {r[0]}（日期：{r[1]}，總名額：{r[2]} 人）" for r in rows])
            return f"⚠️ 無法辨識欲操作的活動。目前系學會舉辦的活動有：\n{list_text}\n請在句子中指明要報名、取消或查詢的活動名稱。"
        finally:
            conn.close()

    # 3. 查詢剩餘名額
    if action == "查詢":
        return query_event(activity)

    # 4. 取消報名
    if action == "取消":
        return cancel_event(cleaned_name, activity)

    # 5. 報名登記
    if action == "報名":
        return register_event(cleaned_name, activity, count)

    return "❓ 系統無法理解您的動作，請嘗試輸入如「我要報名迎新晚會」或「各活動報名狀況」。"


# --- 開始執行流程測試（依序執行 9 步）---
print("=" * 65)
print("🚀 開始執行 9 步業務流程測試...")
print("=" * 65)

# 先重置資料庫，確保環境乾淨
init_database()

workflow_steps = [
    ("步驟 1：正常單人報名", "小明", "我想報名迎新晚會"),
    ("步驟 2：進階攜伴報名（共3人）", "小華", "我要報名籃球友誼賽，算我跟2個朋友一起"),
    ("步驟 3：名額不足測試（已滿3人）", "小強", "我想報名籃球友誼賽"),
    ("步驟 4：重複報名阻擋測試", "小明", "我要報名迎新晚會"),
    ("步驟 5：人數超出 1~3 範圍測試", "大雄", "我要報名程式工作坊，總共4個人"),
    ("步驟 6：報名第 2 個活動（小明）", "小明", "報名程式工作坊"),
    ("步驟 7：超出每人上限 2 場測試", "小明", "報名籃球友誼賽"),
    ("步驟 8：取消報名並釋出名額", "小華", "幫我取消籃球友誼賽"),
    ("步驟 9：查詢全活動報名統計", "", "請給我各活動報名狀況")
]

for title, name, req in workflow_steps:
    print(f"\n▶ 【{title}】")
    print(f"   學生姓名：'{name}' | 輸入需求：'{req}'")
    result = process_system_request(name, req)
    print(f"   回應訊息：{result}")

print("\n" + "=" * 65)
print("📋 流程測試完畢後的最終資料庫狀態：")
print("=" * 65)
show_tables()
````
